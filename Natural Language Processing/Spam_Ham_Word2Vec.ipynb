```python
import re
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
ps=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(0,len(messages)):
  review=re.sub('[^a-z^A-Z]',' ',messages['message'][i])
  review=review.lower()
  review=review.split()
  review=[ps.lemmatize(word) for word in review]
  review=' '.join(review)
  corpus.append(review)
```

```python
words=[]
for sentences in messages['message']:
  sent_token=sent_tokenize(sentences)
  for sent in sent_token:
    words.append(simple_preprocess(sent))
```

```python
words
```

```python
#Train word2vec
model=gensim.models.Word2Vec(words,vector_size=300)
```

```python
#IMP
def averageword_2_vec(doc):
  return np.mean([model.wv[word] for word in doc if word in model.wv.index_to_key])
```

```python
X=[]
for i in tqdm(range(len(words))):
  X.append(averageword_2_vec(words[i]))
```
