# Lemmatization

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemm=WordNetLemmatizer()
```

```python
lemm.lemmatize("programs",pos='v')
```

```python
import nltk
nltk.download('wordnet')
```

```python
words=["writing","write","finally","finalized","programming","program","history"]
```

```python
for word in words:
    print(f"{word} ---> {lemm.lemmatize(word,pos='v')}")
```
