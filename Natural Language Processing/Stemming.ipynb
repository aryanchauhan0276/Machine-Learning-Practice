# Stemming

```python
words=["writing","write","finally","finalized","programming","program","history"]
```

```python
from nltk.stem import PorterStemmer
```

```python
Stemming=PorterStemmer()
```

```python
for word in words:
    print(f"{word} ---> {Stemming.stem(word)}")
```

```python
Stemming.stem("congratulations")
```

# RegexpStemmer()

```python
from nltk.stem import RegexpStemmer
```

```python
RegExp=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
RegExp.stem('eating')
```

```python
RegExp.stem('ingeating')
```

```python
RegExp1=RegexpStemmer('$ing$|s$|e$|able$', min=4)
```

```python
RegExp1.stem('ingeating')
```

# Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
Snowball=SnowballStemmer('english')
```

```python
for word in words:
    print(f"{word} ---> {Snowball.stem(word)}")
```

```python
Stemming.stem('fairly')
```

```python
Snowball.stem('fairly')
```
