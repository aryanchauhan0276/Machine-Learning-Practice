```python
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
import pandas as pd
import numpy as np
import pickle
```

```python
# load the ANN model
model = load_model('model.h5')

# load the encoders & scaler
with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender = pickle.load(file)

with open('Geography_Encoder.pkl', 'rb') as file:
    Geography_Encoder = pickle.load(file)

with open('Scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)
```

```python
input_data={
    'CreditScore':600,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000

}
```

```python
# One-hot encode 'Geography'
geo_encoded = Geography_Encoder.transform([[input_data['Geography']]])
geo_encoded_df = pd.DataFrame(geo_encoded, columns=Geography_Encoder.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_data_df=pd.DataFrame([input_data])
```

```python
input_data=pd.concat([input_data_df,geo_encoded_df],axis=1)
```

```python
input_data.drop(columns=['Geography'],inplace=True)
```

```python
input_data
```

```python
#Encode
input_data['Gender']=label_encoder_gender.transform(input_data['Gender'])
```

```python
input_data
```

```python
## Scaling the input data
input_scaled=scaler.transform(input_data)
```

```python
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_proba = prediction[0][0]
prediction_proba
```

```python
if prediction_proba > 0.5:
    print('The customer is likely to churn.')
else:
    print('The customer is not likely to churn.')
```
