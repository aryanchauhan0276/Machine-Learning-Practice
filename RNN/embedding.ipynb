```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
sent
```

```python
vocab_size=10000
```

```python
one_hot_rep=[one_hot(word,vocab_size)for word in sent]
        
```

```python
one_hot_rep
```

```python
#wordembdding
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_length=8
embed_docs=pad_sequences(one_hot_rep,padding='pre',maxlen=sent_length)
print(embed_docs)
```

```python
#feature_representation
dim=10
```

```python
model=Sequential()
model.add(Embedding(vocab_size,dim,input_length=sent_length))
model.compile(optimizer='adam',loss='mse')
```

```python
model.summary()
```

```python
model.predict(embed_docs)
```
